```python
plt.figure(figsize=(10, 6))

sns.scatterplot(data=data, x='Annual Income (k$)', y='Age', label='Age')

sns.scatterplot(data=data, x='Annual Income (k$)', y='Spending Score (1-100)', label='Spending Score')

plt.title('Annual Income vs Age and Spending Score')
plt.xlabel('Annual Income (k$)')
plt.ylabel('Age / Spending Score')
plt.legend(loc='upper left')
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5), sharex=True)

sns.scatterplot(data=data, x='Annual Income (k$)', y='Age', ax=axes[0])

sns.scatterplot(data=data, x='Annual Income (k$)', y='Spending Score (1-100)', ax=axes[1])

axes[0].set_title('Income vs Age')
axes[1].set_title('Income vs Spending Score')

plt.tight_layout()
plt.show()
```

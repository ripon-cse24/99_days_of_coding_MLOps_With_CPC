```python
plt.figure(figsize=(10, 6))

sns.scatterplot(data=data, x='variance', y='skewness', label='Skewness')

sns.scatterplot(data=data, x='variance', y='curtosis', label='Curtosis')

plt.title('Variance vs Skewness and Curtosis')
plt.xlabel('Variance')
plt.ylabel('Values')
plt.legend()
plt.show()
```

```python
data[['variance', 'skewness', 'curtosis']].corr()
```

```python
plt.figure(figsize=(6, 4))

sns.heatmap(
    data[['variance', 'skewness', 'curtosis']].corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap')
plt.show()
```
